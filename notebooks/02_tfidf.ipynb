```python
from sklearn.feature_extraction.text import CountVectorizer

tiny = ["win free money", "free recharge free", "meet me today"]

cv = CountVectorizer()
counts = cv.fit_transform(tiny)

print(cv.get_feature_names_out())
print(counts.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf = TfidfVectorizer()
scores = tf.fit_transform(tiny)

print(tf.get_feature_names_out())
print(scores.toarray().round(2))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

df = pd.read_csv("../data/SMSSpamCollection", sep="\t", names=["label", "message"])
X_train, X_test, y_train, y_test = train_test_split(
    df["message"], df["label"], test_size=0.2, random_state=42, stratify=df["label"]
)

vectorizer = TfidfVectorizer(lowercase=True, stop_words="english")
# vectorizer = TfidfVectorizer(lowercase=True, stop_words=None,ngram_range=(1, 2),min_df=2)

X_train_tfidf = vectorizer.fit_transform(X_train)   # learn + convert
X_test_tfidf = vectorizer.transform(X_test)         # only convert, no learning

print(X_train_tfidf.shape)
print(X_test_tfidf.shape)
```

```python
import numpy as np

words = vectorizer.get_feature_names_out()

# Average TF-IDF score of each word across spam messages in training data
spam_rows = (y_train == "spam").values
spam_scores = np.asarray(X_train_tfidf[spam_rows].mean(axis=0)).ravel()

top = spam_scores.argsort()[-15:][::-1]
for i in top:
    print(words[i], round(spam_scores[i], 3))
```
